## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the dataset directly via GitHub raw URL
url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# 1. Total records analyzed
print(f"Total analyzed pages: {len(df):,}")

# 2. Correlation between search volume and actual impressions
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Search volume vs impressions correlation: {corr:.3f}")

# 3. Median word count across trend directions
wc = df.groupby("trend_direction")["word_count"].median()
print("\nMedian word count by trend:\n", wc)
```

The numbers reveal two major patterns:
1. Keyword search volume has a near-zero correlation with actual 90-day impressions (~0.01), showing that high search volume does not guarantee traffic.
2. Declining pages and growing pages share nearly identical median word counts, proving that simply adding more words to a page is not the primary lever for ranking retention.
